```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_aot = pd.read_csv("../data/raw/PRSA_Data_Aotizhongxin_20130301-20170228.csv")
df_chang = pd.read_csv("../data/raw/PRSA_Data_Changping_20130301-20170228.csv")
df_ding = pd.read_csv("../data/raw/PRSA_Data_Dingling_20130301-20170228.csv")
```

```python
print(f"Aotizhongxin shape: {df_aot.shape}")
display(df_aot.head())

print(f"Changping shape: {df_chang.shape}")
display(df_chang.head())

print(f"Dingling shape: {df_ding.shape}")
display(df_ding.head())
```

```python
df_combine = pd.concat(
    [df_aot, df_chang, df_ding],
    axis=0,
    ignore_index=True
)

print(f"Combined dataset shape: {df_combine.shape}")
print(f"Value counts for each station:\n{df_combine["station"].value_counts()}")
```

```python
df_combine.info()
```

```python
df_combine["datetime"] = pd.to_datetime(df_combine[["year", "month", "day", "hour"]])
df_combine = df_combine[
    ["No", "datetime", "year", "month", "day", "hour", "PM2.5", "PM10", "SO2", "NO2", "CO", "O3", "TEMP", "PRES", "DEWP", "RAIN", "wd", "WSPM", "station"]
]
df_combine = df_combine.sort_values(by=["station", "datetime"]).reset_index(drop=True)

display(df_combine.head())
```

```python
df_combine = df_combine.drop(["No"], axis=1)

display(df_combine.head())
```

```python
# Create a perfect hourly timeline from start to finish
expected_timeline = pd.date_range(
    start=df_combine["datetime"].min(), 
    end=df_combine["datetime"].max(), 
    freq="h"
)

print(f"Expected total hours per station: {len(expected_timeline)}")

# Check each station against the perfect timeline
for station, group in df_combine.groupby("station"):
    missing_hours = set(expected_timeline) - set(group["datetime"])
    duplicate_hours = group["datetime"].duplicated().sum()
    
    print(f"{station}: {len(missing_hours)} missing timestamps, {duplicate_hours} duplicates")
```

```python
df_combine.isna().sum().sort_values(ascending=False)
```

```python
miss_aot = (df_combine[df_combine["station"] == "Aotizhongxin"].isna().sum()).sort_values(ascending=False).to_frame(name="Aotizhongxin")
miss_chang = (df_combine[df_combine["station"] == "Changping"].isna().sum()).sort_values(ascending=False).to_frame(name="Changping")
miss_ding = (df_combine[df_combine["station"] == "Dingling"].isna().sum()).sort_values(ascending=False).to_frame(name="Dingling")

miss_all = pd.concat([miss_aot, miss_chang, miss_ding], axis=1)

display(miss_all)
```

```python
numerical_columns = ["PM2.5", "PM10", "SO2", "NO2", "CO", "O3", "TEMP", "PRES", "DEWP", "RAIN", "WSPM"]
gap_summary = []

for station, group in df_combine.groupby("station"):
    for col in numerical_columns:
        is_na = group[col].isna()
        gaps = is_na.groupby((~is_na).cumsum()).sum()
        gaps = gaps[gaps > 0]
        
        if len(gaps) > 0:
            gap_summary.append({
                "station": station,
                "column": col,
                "num_gaps": len(gaps),
                "median_gap_hr": gaps.median(),
                "mean_gap_hr": round(gaps.mean(), 2),
                "max_gap_hr": gaps.max(),
                "gaps_over_6h": (gaps > 6).sum()
            })

df_gap_summary = pd.DataFrame(gap_summary)
display(df_gap_summary.pivot(index="column", columns="station", values=["max_gap_hr", "gaps_over_6h", "num_gaps"]))
```

```python
df_clean = df_combine.copy()
df_clean[numerical_columns] = df_clean.groupby("station")[numerical_columns].transform(lambda x: x.interpolate("linear", limit=6, limit_area="inside"))
df_clean["wd"] = df_clean.groupby("station")["wd"].ffill(limit=6)

df_clean.isna().sum().sort_values(ascending=False)
```
